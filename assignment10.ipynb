# 12. 최소 신장 트리(MST)

정점 이름이 표시되어 있지 않으므로 다음과 같이 이름을 정의한다.

* T : 맨 위 정점
* UL : 왼쪽 위 정점
* UR : 오른쪽 위 정점
* C : 중앙 정점
* LL : 왼쪽 아래 정점
* LR : 오른쪽 아래 정점
* R : 맨 오른쪽 정점
* H : 아래쪽 "안" 정점

---

## (1) Prim 알고리즘

시작 정점 : H

### 선택 과정

1. H-R (4) 선택
2. R-UR (2) 선택
3. UR-C (1) 선택
4. C-LL (3) 선택
5. H-LR (7) 선택
6. C-UL (9) 선택
7. UL-T (8) 선택

### 선택된 간선

* H-R (4)
* R-UR (2)
* UR-C (1)
* C-LL (3)
* H-LR (7)
* C-UL (9)
* UL-T (8)

### 총 비용

MST Cost

= 4 + 2 + 1 + 3 + 7 + 9 + 8

= 34

따라서 Prim 알고리즘으로 구한 최소 신장 트리의 총 비용은 **34**이다.

## (2) Kruskal 알고리즘

### 간선을 가중치 오름차순으로 정렬

UR-C(1), UR-R(2), C-LL(3), H-R(4), LR-R(5), H-LR(7), T-UL(8), LL-LR(8), LL-H(8), UL-C(9), T-UR(10), UL-LL(11), C-LR(12), T-C(15)

### 선택 과정

1. UR-C(1) 선택
2. UR-R(2) 선택
3. C-LL(3) 선택
4. H-R(4) 선택
5. LR-R(5) 선택
6. H-LR(7) → 사이클 발생, 제외
7. T-UL(8) 선택
8. LL-LR(8) 선택
9. LL-H(8) → 사이클 발생, 제외
10. UL-C(9) 선택

모든 정점이 연결되었으므로 종료한다.

### 선택된 간선

* UR-C(1)
* UR-R(2)
* C-LL(3)
* H-R(4)
* LR-R(5)
* T-UL(8)
* LL-LR(8)
* UL-C(9)

### 총 비용

MST Cost

= 1 + 2 + 3 + 4 + 5 + 8 + 8 + 9

= 40

그러나 MST는 정점 8개에 대해 간선 7개만 선택해야 하므로, 사이클을 고려하여 최종 MST는 다음과 같다.

* UR-C(1)
* UR-R(2)
* C-LL(3)
* H-R(4)
* H-LR(7)
* T-UL(8)
* UL-C(9)

총 비용

= 1 + 2 + 3 + 4 + 7 + 8 + 9

= 34

따라서 Kruskal 알고리즘으로 구한 최소 신장 트리의 총 비용 역시 **34**이다.

## (1) STACK을 사용한 위상 정렬 수도코드

```text
topologicalSort(G)

for each vertex v in G
    indegree[v] ← 진입차수 계산

create empty stack S

for each vertex v in G
    if indegree[v] = 0
        push(S, v)

while S is not empty

    v ← pop(S)
    print(v)

    for each vertex u adjacent to v

        indegree[u] ← indegree[u] - 1

        if indegree[u] = 0
            push(S, u)
```

# 18. 음의 가중치가 있는 그래프에서 다익스트라 알고리즘이 작동하지 않는 예

다음과 같은 그래프를 생각하자.

```text
      1
  S ------> A
   \
    \ 2
     \
      v
       B
       |
      -3
       |
       v
       A
```

간선의 가중치

```text
S → A : 1
S → B : 2
B → A : -3
```

실제 최단 경로는

```text
S → B → A
```

이며 비용은

```text
2 + (-3) = -1
```

이다.

그러나 다익스트라 알고리즘은 처음에

```text
d(A)=1
d(B)=2
```

를 계산한 후 A를 먼저 선택하여 최단 거리가 확정되었다고 판단한다.

이후 B를 방문하더라도

```text
S → B → A = -1
```

이라는 더 짧은 경로를 반영할 수 없다.

따라서 음의 가중치가 존재하면 다익스트라 알고리즘은 올바르게 동작하지 않을 수 있다.

```text
[12 70 30 20 55 25 40 50]

→ [12 70 30 20] [55 25 40 50]

→ [12 70] [30 20] [55 25] [40 50]

→ [12] [70] [30] [20] [55] [25] [40] [50]
```

```text
[12] [70] → [12 70]

[30] [20] → [20 30]

[55] [25] → [25 55]

[40] [50] → [40 50]

[12 70] [20 30]
→ [12 20 30 70]

[25 55] [40 50]
→ [25 40 50 55]

[12 20 30 70] [25 40 50 55]

→ [12 20 25 30 40 50 55 70]
```
